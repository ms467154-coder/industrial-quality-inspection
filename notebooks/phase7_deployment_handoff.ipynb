# Phase 7 — Deployment Handoff
Package prediction, confidence, and Grad-CAM evidence for human review.

```python
import json, pandas as pd
metrics=json.load(open('phase4_complete_metrics.json'))
pred=pd.read_csv('phase4_test_predictions.csv')
cases=pd.read_csv('phase6_gradcam_cases.csv')
print(metrics['roc_auc'], len(pred), len(cases))
```

## Review policy
Grad-CAM is decision support: inspectors should confirm the highlighted region before accepting or rejecting a casting.

```python
summary={'test_images':len(pred),'false_positives':int((pred.case=='False positive').sum()),'false_negatives':int((pred.case=='False negative').sum()),'explanations':len(cases)}
with open('phase7_review_summary.json','w') as f: json.dump(summary,f,indent=2)
print(summary)
print('Phase 7 deployment handoff ready; human validation remains required.')
```
