```python
IMG_SIZE = (224, 224)
label_id = {'OK': 0, 'Defective': 1}
def load_image(path):
    with Image.open(path) as im: im = im.convert('RGB').resize(IMG_SIZE)
    return np.asarray(im, dtype=np.float32) / 255.0
print('Phase 3 loader contract ready; model training not started.')
```
