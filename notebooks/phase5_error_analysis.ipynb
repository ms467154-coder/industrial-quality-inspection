# Phase 5 — Error Analysis Handoff
Use actual Phase 4 predictions to inspect operating thresholds and errors.

```python
import numpy as np, pandas as pd
from sklearn.metrics import precision_recall_fscore_support
pred = pd.read_csv('phase4_test_predictions.csv')
thresholds = np.arange(.10, .91, .05)
print(f'Predictions loaded: {len(pred):,}')
```

```python
rows = []
for t in thresholds:
    guess = (pred.defective_score >= t).astype(int)
    p,r,f,_ = precision_recall_fscore_support(pred.actual,guess,average='binary',zero_division=0)
    rows.append({'threshold':t,'precision':p,'recall':r,'f1':f})
threshold_table = pd.DataFrame(rows); display(threshold_table)
threshold_table.to_csv('phase5_threshold_analysis.csv',index=False)
```
