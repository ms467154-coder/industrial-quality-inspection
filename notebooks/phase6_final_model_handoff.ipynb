# Phase 6 — Final Model Handoff
Select the measured best model and preserve its evaluation record.

```python
import json, shutil
import pandas as pd
comparison=pd.read_csv('phase5_model_comparison.csv')
best=comparison.iloc[comparison.F1.argmax()]
print(best.to_dict())
```

```python
source='phase3_mobilenetv2_head_baseline.pt' if best.Model.startswith('Original') else ('phase5_112px_unweighted_head.pt' if 'unweighted' in best.Model else 'phase5_112px_class_weighted_head.pt')
shutil.copyfile(source,'phase6_selected_model.pt')
with open('phase6_selected_model.json','w') as f: json.dump({'source_checkpoint':source,**best.to_dict()},f,indent=2)
print(f'Phase 6 selected: {source}')
print('Phase 6 handoff ready; no additional training was run.')
```
